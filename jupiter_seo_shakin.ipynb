```python
df_week = pd.DataFrame()

for single_date in daterange[:3]:
    params_listQueries = dict(
        ids=COUNTER_ID,
        metrics="ym:s:visits",
        dimensions="ym:s:startURL,ym:s:Date",
        date1 = single_date.strftime("%Y-%m-%d"),
        date2 = single_date.strftime("%Y-%m-%d"),
        limit=100000,
        receive_all_data=True,
        accuracy=1
        )
    while True:
        try:
            result = api.stats().get(params=params_listQueries)
            data = result().transform()
            data[0]
            df = pd.DataFrame(data, columns = data[0])[1:]
            df_week = pd.concat([df_week, df])
            time.sleep(5)
        except Exception:
            continue
        else:
            break
```
